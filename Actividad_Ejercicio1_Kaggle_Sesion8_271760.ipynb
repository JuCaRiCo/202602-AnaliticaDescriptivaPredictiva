**Paso 3 — Cargar el archivo descargado en un DataFrame.**

```python
import glob
import pandas as pd

archivos_csv = glob.glob("datos_drug/*.csv")
print(archivos_csv)

# Completa: carga el primer archivo de archivos_csv en un DataFrame llamado df_drug
if archivos_csv:
    df_drug = pd.read_csv(archivos_csv[0])
```
