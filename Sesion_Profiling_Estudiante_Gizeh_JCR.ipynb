## Parte 1 — Funciones a nivel Dataset

Cada función recibe el DataFrame completo (`df`) y regresa **un solo valor**. Completa el cuerpo de cada función donde dice `# TODO`.

```python
def n_columnas(df):
    """Regresa el número de variables (columnas) del DataFrame."""
    return df.shape[1]


def n_filas(df):
    """Regresa el número de observaciones (filas) del DataFrame."""
    return df.shape[0]
```

```python
def n_datos_ausentes(df):
    """Regresa el número total de celdas con datos ausentes (NaN) en todo el DataFrame."""
    return df.isna().sum().sum()

def pct_celdas_ausentes(df):
    """Regresa el porcentaje de celdas con datos ausentes sobre el total de celdas
    (filas x columnas) del DataFrame."""
    totalCeldas = len(df) * len(df.columns)

    return (n_datos_ausentes(df) / totalCeldas) * 100
```

```python
def n_filas_duplicadas(df):
    """Regresa el número de filas duplicadas en el DataFrame."""
    return df.duplicated().sum()

def pct_filas_duplicadas(df):
    """Regresa el porcentaje de filas duplicadas sobre el total de filas."""
    return (n_filas_duplicadas(df) / len(df)) * 100
```

```python
def n_variables_numericas(df):
    """Regresa el número de columnas con tipo de dato numérico (int o float)."""
    return len(df.select_dtypes(include=['number']).columns)

def n_variables_booleanas(df):
    """Regresa el número de columnas con tipo de dato booleano."""
    return len(df.select_dtypes(include=['bool']).columns)

def n_variables_categoricas(df):
    """Regresa el número de columnas con tipo de dato object o category."""
    return len(df.select_dtypes(include=['object', 'category']).columns)
```

### Verifica tus funciones

Compara lo que regresan tus funciones contra `.info()` y `.describe()` — deben coincidir.

```python
print('Columnas:', n_columnas(df))
print('Filas:', n_filas(df))
print('Celdas ausentes:', n_datos_ausentes(df))
print(f'% celdas ausentes: {pct_celdas_ausentes(df):.2f}%')
print('Filas duplicadas:', n_filas_duplicadas(df))
print(f'% filas duplicadas: {pct_filas_duplicadas(df):.2f}%')
print('Variables numéricas:', n_variables_numericas(df))
print('Variables booleanas:', n_variables_booleanas(df))
print('Variables categóricas:', n_variables_categoricas(df))

print()
df.info()
print()
df.describe()
```

### Función resumen

Combina todas las funciones anteriores en un solo diccionario — tu propio "reporte ejecutivo" del dataset.

```python
def resumen_dataset(df):
    """Regresa un diccionario con el resumen general del DataFrame,
    combinando todas las funciones de la Parte 1."""
    return {
        'n_columnas': n_columnas(df),
        'n_filas': n_filas(df),
        'n_datos_ausentes': n_datos_ausentes(df),
        'pct_celdas_ausentes': pct_celdas_ausentes(df),
        'n_filas_duplicadas': n_filas_duplicadas(df),
        'pct_filas_duplicadas': pct_filas_duplicadas(df),
        'n_variables_numericas': n_variables_numericas(df),
        'n_variables_booleanas': n_variables_booleanas(df),
        'n_variables_categoricas': n_variables_categoricas(df),
    }

resumen_dataset(df)
```

## Parte 2 — Funciones a nivel Columna

Cada función recibe el DataFrame y el **nombre de una columna** (`col`), y regresa un solo valor para esa columna.

```python
def total_ausentes(df, col):
    """Regresa el número de valores ausentes (NaN) en la columna col."""
    return df[col].isna().sum()

def pct_ausentes(df, col):
    """Regresa el porcentaje de valores ausentes en la columna col."""
    return (total_ausentes(df, col) / len(df)) * 100
```

```python
def media(df, col):
    """Regresa la media de la columna col (asume que es numérica)."""
    return df[col].mean()

def minimo(df, col):
    """Regresa el valor mínimo de la columna col."""
    return df[col].min()

def maximo(df, col):
    """Regresa el valor máximo de la columna col."""
    return df[col].max()
```

```python
def total_ceros(df, col):
    """Regresa el número de valores iguales a 0 en la columna col."""
    return (df[col] == 0).sum()

def pct_ceros(df, col):
    """Regresa el porcentaje de valores iguales a 0 en la columna col."""
    return (total_ceros(df, col) / len(df)) * 100
```

```python
def total_negativos(df, col):
    """Regresa el número de valores negativos en la columna col."""
    return (df[col] < 0).sum()

def pct_negativos(df, col):
    """Regresa el porcentaje de valores negativos en la columna col."""
    return (total_negativos(df, col) / len(df)) * 100
```

### Función resumen

```python
def resumen_columna(df, col):
    """Regresa un diccionario con el resumen de la columna col,
    combinando todas las funciones de la Parte 2."""
    return {
        'total_ausentes': total_ausentes(df, col),
        'pct_ausentes': pct_ausentes(df, col),
        'media': media(df, col),
        'minimo': minimo(df, col),
        'maximo': maximo(df, col),
        'total_ceros': total_ceros(df, col),
        'pct_ceros': pct_ceros(df, col),
        'total_negativos': total_negativos(df, col),
        'pct_negativos': pct_negativos(df, col),
    }

print('Rank:', resumen_columna(df, 'Rank'))
print('Year:',resumen_columna(df, 'Year'))
print('NA_Sales:', resumen_columna(df, 'NA_Sales'))
print('EU_Sales:',resumen_columna(df, 'EU_Sales'))
print('JP_Sales:',resumen_columna(df, 'JP_Sales'))
print('Other_Sales:',resumen_columna(df, 'Other_Sales'))
print('Global_Sales:', resumen_columna(df, 'Global_Sales'))
```
