# 1. Operaciones básicas

Instrucciones:
1. Crea un array de 10 elementos con valores entre 1 y 10.
2. Calcula la media, mediana, y desviación estándar de los valores.
3. Multiplica cada valor por 2.

```python
import numpy as np
import pandas as pd

#1. Creando el array
array1to10 = np.arange(1,11)

#2. Calculo de la media, mediana, y desviación estándar
media = np.mean(array1to10)
mediana = np.median(array1to10)
desviacion_estandar = np.std(array1to10)

print('Array original: ', array1to10)
print('Media: ', media)
print('Mediana: ', mediana)
print('Desviación estándar: ', desviacion_estandar)

#3. Multiplicando por 2
array1to10by2 = array1to10 * 2
print('Multiplicado por 2: ', array1to10by2)

```

Observaciones:

- La media y la mediana son iguales debido a su distribucion simetrica  de la secuencia del 1 al 10.

# 2. Operaciones con DataFrames
Instrucciones:
1. Crea un DataFrame con 3 columnas (A, B, C) y 5 filas con números
aleatorios enteros.
2. Renombra las columnas a Col1, Col2, Col3.
3. Añade una nueva columna que sea la suma de las columnas Col1 y Col2.
4. Elimina todas las filas donde el valor de Col3 sea menor que 10.

```python
#Se fija la semilla para la generacion de numeros random
np.random.seed(42)

#1. Generando el DataFrame, definiendo valores minimos y maximos, asi como el tamaño del array
df = pd.DataFrame(np.random.randint(1, 20, size=(5, 3)), columns=['A', 'B', 'C'])
print("DataFrame original:\n", df, "\n")

#2. Renombrando las columnas
df.columns = ['Col1', 'Col2', 'Col3']
print("DataFrame con columnas renombradas:\n", df, "\n")

#3. Sumando col1, col2 y agregando a columna
df['Suma'] = df['Col1'] + df['Col2']
print('DataFrame con nueva columna\n', df)

#4. Eliminando records donde Col3 < 10
dfNuevo = df[df['Col3'] >= 10].reset_index(drop=True)
print('DataFrame con Col3 >= 10\n', dfNuevo)
```

# Observaciones:
Los registros se eliminaron satisfactoriamente tras la evaluacion pedida.
La suma es directa y simple entre columnas para generar una adicional

# 3. Estadística descriptiva
Objetivo: Aplicar análisis estadístico básico.
Instrucciones:
1. Crea un DataFrame con 100 valores aleatorios entre 0 y 50.
2. Calcula la media, mediana y moda de los valores.
3. Encuentra el valor mínimo y el valor máximo del DataFrame.

```python
#1. Creando DataFrame
valores31 = np.random.randint(0, 51, size=(100))
df31 = pd.DataFrame({'valores':valores31})
print(df31)

#2. calculando la media, mediana y moda
media31 = df31['valores'].mean()
print('Media: ', media31)
mediana31 = df31['valores'].median()
print(  'Mediana: ', mediana31)
moda31 = df31['valores'].mode().iloc[0]
print('Moda: ', moda31 )

#3. Minimo y Maximo del df
print('Minimo: ', df31['valores'].min())
print('Maximo: ', df31['valores'].max())


```

Realizar este tipo de calculos es sumamente facil pandas
