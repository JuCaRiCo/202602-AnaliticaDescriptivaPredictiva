```python
import numpy as np
import matplotlib.pyplot as plt

def CalculoTiempoTotal(B, N_total, t0, c):
    return (N_total * t0) / B + c * (B**2)

# Configurar escenario
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- Caso A ---
N_A, t0_A, c_A = 10000, 0.005, 1e-5
B_opt_A = (N_A * t0_A / (2 * c_A))**(1/3)
T_opt_A = CalculoTiempoTotal(B_opt_A, N_A, t0_A, c_A)

B_vals_A = np.linspace(20, 400, 500)
axes[0].plot(B_vals_A, CalculoTiempoTotal(B_vals_A, N_A, t0_A, c_A), label="$T(B)$", color='blue', lw=2)
axes[0].plot(B_opt_A, T_opt_A, 'ro', markersize=8, label=f'Mínimo: $B^* \\approx {B_opt_A:.1f}$')
axes[0].axvline(B_opt_A, color='red', linestyle='--', alpha=0.6)
axes[0].set_title("Caso A: Dataset Pequeño ($N=10^4$, $t_0=5ms$, $c=10^{-5}$)")
axes[0].set_xlabel("Tamaño de Lote ($B$)")
axes[0].set_ylabel("Tiempo Total $T(B)$ (segundos)")
axes[0].grid(True, linestyle=':', alpha=0.7)
axes[0].legend()

# --- Caso B ---
N_B, t0_B, c_B = 1000000, 0.1, 1e-4
B_opt_B = (N_B * t0_B / (2 * c_B))**(1/3)
T_opt_B = CalculoTiempoTotal(B_opt_B, N_B, t0_B, c_B)

B_vals_B = np.linspace(100, 2500, 500)
axes[1].plot(B_vals_B, CalculoTiempoTotal(B_vals_B, N_B, t0_B, c_B), label="$T(B)$", color='green', lw=2)
axes[1].plot(B_opt_B, T_opt_B, 'ro', markersize=8, label=f'Mínimo: $B^* \\approx {B_opt_B:.1f}$')
axes[1].axvline(B_opt_B, color='red', linestyle='--', alpha=0.6)
axes[1].set_title("Caso B: Dataset Mediano ($N=10^6$, $t_0=0.1s$, $c=10^{-4}$)")
axes[1].set_xlabel("Tamaño de Lote ($B$)")
axes[1].set_ylabel("Tiempo Total $T(B)$ (segundos)")
axes[1].grid(True, linestyle=':', alpha=0.7)
axes[1].legend()

plt.tight_layout()
plt.show()
```
