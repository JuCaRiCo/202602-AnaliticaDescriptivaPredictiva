## Actividad 1. Diagnóstico inicial (20 puntos)

### 1.1 Conteo de duplicados exactos (5 pts)

Calcula el número de filas marcadas como duplicadas en `hoteles` considerando **todas** las columnas, con `keep='first'`, `keep='last'` y `keep=False`. Imprime los tres resultados con un texto que explique qué significa cada uno.

```python

#Total de records existentes
totalRecords = len(hoteles)

#Total de records duplicados usando 'first'
duplicatedFirst = hoteles.duplicated(keep='first').sum()
print("Duplicados (keep='first'): ", duplicatedFirst)

#Total de records duplicados usando 'last'
duplicatedLast = hoteles.duplicated(keep='last').sum()
print("Duplicados (keep='last'): ", duplicatedLast)

#Total de records duplicados usando 'False'
duplicatedFalse = hoteles.duplicated(keep=False).sum()
print('Duplicados (keep=False): ', duplicatedFalse)
print('')
print("Explicación de significados:")
print("keep='first' (", duplicatedFirst, "): hace referencia a los records repetidos conservando el primer record ")
print("keep='last' (", duplicatedLast, "): hace referencia a los records repetidos conservando el ultimo record")
print("keep=False (", duplicatedFalse, "): hace referencia a los records que no son unicos")
```

### 1.2 Magnitud del problema (5 pts)

Con los resultados de 1.1, calcula e imprime:

- El **porcentaje** de filas sobrantes respecto al total.
- El número de **grupos repetidos** (combinaciones de valores que aparecen más de una vez).

```python
# sacamos el porcentaje dividiendo duplicados entre total de records, y multiplicamos por 100,
# haciendo round a 2 decimales
porcentajeSobrantes = round(duplicatedFirst / totalRecords * 100, 2)
print('Porcentaje de records sobrantes respecto al total: ', porcentajeSobrantes, '%')

# sacamos los duplicados y lo usamos como subconjunto de hoteles y aplicamos drop_duplicates
# para remover el total duplicados y dejar solo un registro de los duplicados
gruposRepetidos = len(hoteles[hoteles.duplicated(keep=False)].drop_duplicates())
print('Numero de grupos repetidos: ', gruposRepetidos)
```

### 1.3 Duplicados por hotel (5 pts)

Determina **cuántas** filas sobrantes (`keep='first'`) hay en cada hotel y qué **porcentaje** representan respecto al total de reservas de ese hotel.

Sugerencia: crea una copia de `hoteles`, agrega una columna booleana `es_duplicado` con el resultado de `duplicated()` y agrupa por `hotel`.

```python
hotelBkp = hoteles.copy()

# se crea una columna para guardar el valor booleando para saber si es duplicado
hotelBkp['es_duplicado'] = hoteles.duplicated(keep='first')

#Resumen Total por Hotel
totalPorHotel = hotelBkp.groupby('hotel')['es_duplicado'].count()
#Resumen de Sobrantes por hotel
sobrantesPorHotel = hotelBkp.groupby('hotel')['es_duplicado'].sum()
#Porcentaje sobrantes sobre total por hotel
porcentajePorHotel = round(sobrantesPorHotel / totalPorHotel * 100, 2)

print('Total Por Hotel:\n',totalPorHotel)
print('\nSobrantes Por Hotel:\n',sobrantesPorHotel)
print('\nPorcentaje Sobrantes Por Hotel:\n',porcentajePorHotel)
```
