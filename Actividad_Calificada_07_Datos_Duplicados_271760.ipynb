## Actividad 1. Diagnóstico inicial (20 puntos)

### 1.1 Conteo de duplicados exactos (5 pts)

Calcula el número de filas marcadas como duplicadas en `hoteles` considerando **todas** las columnas, con `keep='first'`, `keep='last'` y `keep=False`. Imprime los tres resultados con un texto que explique qué significa cada uno.

```python

#Total de records existentes
totalRecords = len(hoteles)

#Total de records duplicados usando 'first'
duplicatedFirst = hoteles.duplicated(keep='first').sum()
print("Duplicados (keep='first'): ", duplicatedFirst)

#Total de records duplicados usando 'last'
duplicatedLast = hoteles.duplicated(keep='last').sum()
print("Duplicados (keep='last'): ", duplicatedLast)

#Total de records duplicados usando 'False'
duplicatedFalse = hoteles.duplicated(keep=False).sum()
print('Duplicados (keep=False): ', duplicatedFalse)
print('')
print("Explicación de significados:")
print("keep='first' (", duplicatedFirst, "): hace referencia a los records repetidos conservando el primer record ")
print("keep='last' (", duplicatedLast, "): hace referencia a los records repetidos conservando el ultimo record")
print("keep=False (", duplicatedFalse, "): hace referencia a los records que no son unicos")
```

### 1.2 Magnitud del problema (5 pts)

Con los resultados de 1.1, calcula e imprime:

- El **porcentaje** de filas sobrantes respecto al total.
- El número de **grupos repetidos** (combinaciones de valores que aparecen más de una vez).

```python
# sacamos el porcentaje dividiendo duplicados entre total de records, y multiplicamos por 100,
# haciendo round a 2 decimales
porcentajeSobrantes = round(duplicatedFirst / totalRecords * 100, 2)
print('Porcentaje de records sobrantes respecto al total: ', porcentajeSobrantes, '%')

# sacamos los duplicados y lo usamos como subconjunto de hoteles y aplicamos drop_duplicates
# para remover el total duplicados y dejar solo un registro de los duplicados
gruposRepetidos = len(hoteles[hoteles.duplicated(keep=False)].drop_duplicates())
print('Numero de grupos repetidos: ', gruposRepetidos)
```

### 1.3 Duplicados por hotel (5 pts)

Determina **cuántas** filas sobrantes (`keep='first'`) hay en cada hotel y qué **porcentaje** representan respecto al total de reservas de ese hotel.

Sugerencia: crea una copia de `hoteles`, agrega una columna booleana `es_duplicado` con el resultado de `duplicated()` y agrupa por `hotel`.

```python
hotelBkp = hoteles.copy()

# se crea una columna para guardar el valor booleando para saber si es duplicado
hotelBkp['es_duplicado'] = hoteles.duplicated(keep='first')

#Resumen Total por Hotel
totalPorHotel = hotelBkp.groupby('hotel')['es_duplicado'].count()
#Resumen de Sobrantes por hotel
sobrantesPorHotel = hotelBkp.groupby('hotel')['es_duplicado'].sum()
#Porcentaje sobrantes sobre total por hotel
porcentajePorHotel = round(sobrantesPorHotel / totalPorHotel * 100, 2)

print('Total Por Hotel:\n',totalPorHotel)
print('\nSobrantes Por Hotel:\n',sobrantesPorHotel)
print('\nPorcentaje Sobrantes Por Hotel:\n',porcentajePorHotel)
```

### 3.1 Versión sin duplicados (5 pts)

Crea `hoteles_limpio` eliminando los duplicados exactos, conservando la primera aparición y **renumerando el índice**. Imprime el número de filas antes y después y cuántas se eliminaron.

### 3.2 Métricas por hotel (8 pts)

Para `hoteles` y para `hoteles_limpio`, calcula por hotel con `groupby()`:

- Número de reservas (`.size()`).
- Tasa de cancelación (media de `is_canceled`).
- Tarifa diaria promedio (media de `adr`).

Nota: para calcular la media de varias columnas a la vez, selecciona una lista de columnas: `df.groupby('hotel')[['col1', 'col2']].mean()`.

Guarda las medias en dos `DataFrame`, `resumen_original` y `resumen_limpio`, con `hotel` como columna (usa `reset_index()`).

```python
print('Numero de Reservas: ')
cantOriginal = hoteles.groupby('hotel').size().reset_index()
cantLimpio = hotelesLimpio.groupby('hotel').size().reset_index()

print('Cantidad Original: ', cantOriginal)
print('\nCantidad Limpia: ', cantLimpio)

print('\n---------------------------------------------------------------')
print('Tasa de Cancelacion (media de is_canceled)')
mediasOriginal = hoteles.groupby('hotel')[['is_canceled', 'adr']].mean().reset_index()
mediasLimpio = hotelesLimpio.groupby('hotel')[['is_canceled', 'adr']].mean().reset_index()

print('Media Original: ', mediasOriginal)
print('\nMedia Limpia: ', mediasLimpio)

print('\n---------------------------------------------------------------')
print('Tarifa diaria promedio')
resumen_original = pd.merge(cantOriginal, mediasOriginal, on='hotel')
resumen_limpio = pd.merge(cantLimpio, mediasLimpio, on='hotel')

print('Resumen Original: ', resumen_original)
print('\nResumen Limpia: ', resumen_limpio)
```

```python
# Creamos copias explicitas para agregar la columna 'version' a cada resumen
resumenOriginalV = resumen_original.copy()
resumenOriginalV['version'] = 'original'

resumenLimpioV = resumen_limpio.copy()
resumenLimpioV['version'] = 'sin duplicados'

print('\n---------------------------------------------------------------')
comparacion = pd.concat([resumenOriginalV, resumenLimpioV], ignore_index=True)

# Ordenamos por hotel, version
comparacion = comparacion.sort_values(by=['hotel', 'version']).reset_index(drop=True)

comparacion

```

### 4.2 Duplicados parciales (10 pts)

La columna `reservation_status_date` registra la fecha del último cambio de estado. Detecta las filas de `hoteles_limpio` que **coinciden en todas las columnas excepto `reservation_status_date`**.

1. Cuenta cuántas filas sobrantes hay con ese criterio.

Después responde: ¿qué podrían representar estas filas? ¿Las eliminarías? Justifica.

```python
# Obtenemos la lista de columnas excluyendo 'reservation_status_date'
colsSinFecha = [col for col in hoteles_limpio.columns if col != 'reservation_status_date']

# Identificamos los duplicados (keep='first')
duplicadosParcialesSobrantes = hoteles_limpio.duplicated(subset=colsSinFecha, keep='first').sum()
print('Duplicados Parciales Sobrantes: ', duplicadosParcialesSobrantes)

# Filtramos las filas involucradas en este tipo de duplicado (keep=False)
recordsParciales = hoteles_limpio[hoteles_limpio.duplicated(subset=colsSinFecha, keep=False)]
print('Filtrado de duplicados: \n',recordsParciales)

colsParaMostrar = [
    'hotel',
    'lead_time',
    'arrival_date_year',
    'arrival_date_month',
    'arrival_date_day_of_month',
    'adr',
    'reservation_status',
    'reservation_status_date',
]
colsParaOrden = [
    'hotel',
    'arrival_date_year',
    'arrival_date_month',
    'arrival_date_day_of_month',
    'lead_time',
]

parcialesOrdenados = recordsParciales.sort_values(by=colsParaOrden)
print('\nRecords con duplicacion parcial\n', parcialesOrdenados[colsParaMostrar].head(10).to_string())

# Distribucion de reservation_status en estas filas
distStatus = parcialesOrdenados['reservation_status'].value_counts(normalize=True) * 100
print('\nDistribución de reservation_status en duplicados parciales: \n', distStatus.round(2))
```
