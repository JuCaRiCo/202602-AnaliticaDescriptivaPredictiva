Cargamos el archivo desde Google Drive e iteramos los renglones para obtener nombre de la empresa y costo de la accion

```python
with open('/content/drive/MyDrive/MIAAD/202602-AnaliticaDescriptivaPredictiva/informe_acciones.txt', 'r', encoding='latin-1') as informeAcciones:
  companyName = ''
  stockValue = 0.0

  # Se itera el archivo, linea por linea
  for line in informeAcciones:
      line = line.rstrip()

      if line.startswith('Asunto: Informe de Acción - '):
        #Obtenemos una lista separando por ' - ' y tomamos el segundo elemento
        companyName = line.split(' - ')[1]
      elif line.startswith('Valor actual'):
        #Obtenemos el valor actual de la accion
        stockValue = re.search(r'(\$[0-9.]+)', line)

        print('Empresa:', companyName, '\nValor actual:', stockValue.group(1))
        print('-------------------------')
```
