```python
import numpy as np
# Número de simulaciones
N = 1000000

# ----- Simulacion con los datos del Ejercicio -----
dado1 = np.random.randint(1, 7, size=N)
dado2 = np.random.randint(1, 7, size=N)
sumas = dado1 + dado2
prob_suma_7 = np.sum(sumas == 7) / N
prob_suma_mayor_10 = np.sum(sumas > 10) / N
print(f"--- Resultados para Ejercicio 5 ---")
print(f"Probabilidad empírica (Suma = 7): {prob_suma_7:.4f} (Teórica: 0.1667)")
print(f"Probabilidad empírica (Suma > 10): {prob_suma_mayor_10:.4f} (Teórica: 0.0833)")
```
