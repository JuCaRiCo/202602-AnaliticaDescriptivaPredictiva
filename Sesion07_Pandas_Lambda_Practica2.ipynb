**Agrupar datos**

Ejercicio: Si tuviéramos datos de diferentes cursos, puedes agrupar por curso y obtener la nota promedio en cada asignatura.

```python
# Crear DataFrame con curso
data_con_curso = {
    'Estudiante': ['Juan', 'María', 'Pedro', 'Ana', 'Luis'],
    'Curso': ['A', 'A', 'B', 'B', 'A'],
    'Matemáticas': [8, 9, 6, 7, 8],
    'Lengua': [7, 9, 8, 6, 8]
}
df_curso = pd.DataFrame(data_con_curso)

print(df_curso)

# Agrupar por curso y calcular el promedio por asignatura solo para las columnas numéricas
promedios_curso = df_curso.groupby('Curso').agg({
    'Matemáticas': 'mean',
    'Lengua': 'mean'
})

# Mostrar los promedios por curso
print(promedios_curso)
```

Encontrar estudiantes con alguna asignatura reprobada (nota menor a 5).

Ejercicio: Filtra los estudiantes que tengan al menos una asignatura con una nota menor a 5.

```python
# Leer el archivo CSV
# La separación por default es coma (,)
df = pd.read_csv('/content/drive/MyDrive/ClassFiles/estudiantes.csv')

# Mostrar las primeras 5 filas
print(df.head())
```

```python
# Filtrar estudiantes con alguna asignatura reprobada
estudiantes_reprobados = df[(df['Matemáticas'] < 5) | (df['Lengua'] < 5) | (df['Ciencias'] < 5)]

# Mostrar los estudiantes reprobados
print(estudiantes_reprobados[['Estudiante', 'Matemáticas', 'Lengua', 'Ciencias']])
```

```python

# Asegurarnos de que las columnas de notas son numéricas (si hay algún error en el tipo de dato, convertimos a numérico)
df[['Matemáticas', 'Lengua', 'Ciencias']] = df[['Matemáticas', 'Lengua', 'Ciencias']].apply(pd.to_numeric, errors='coerce')

# Agrupar por 'Curso' y calcular el promedio por asignatura
promedios_curso = df.groupby('Curso')[['Matemáticas', 'Lengua', 'Ciencias']].mean()

# Mostrar los promedios por curso
print(promedios_curso)
```

## Bloque 3 — lambda con `apply()` y `map()`

### Reconexión con Sesión 5

En Sesión 5 usaron lambda como argumento de `sorted()`. lambda sigue siendo lo mismo: una función sin nombre, de una sola expresión. Lo que cambia hoy es quién la invoca.

```python
lista_estudiantes = estudiantes.to_dict('records')
ordenados = sorted(lista_estudiantes, key=lambda e: e['calificacion'])
ordenados
```

### `Series.map()`: transformar valor por valor

**Predicción:** ¿qué contiene cada elemento del resultado?

```python
estudiantes['calificacion'].map(lambda x: x / 10)
```

### `.apply()` con una función nombrada

Empezamos con `def` (ya conocida desde Sesión 5) para ver con claridad qué hace `apply()`: toma cada valor de la columna y lo pasa como argumento a la función.

```python
def categorizar(nota):
    if nota >= 90:
        return 'Excelente'
    elif nota >= 80:
        return 'Bueno'
    return 'Regular'

estudiantes['calificacion'].apply(categorizar)
```

### La misma idea, con lambda

Cuando la función es una sola expresión corta, lambda evita tener que definirla aparte con `def`.

```python
estudiantes['aprobado'] = estudiantes['calificacion'].apply(
    lambda nota: 'Aprobado' if nota >= 80 else 'Reprobado'
)
estudiantes
```

### `DataFrame.apply()` por fila: `axis=1`

Cuando la función necesita varias columnas de la misma fila a la vez (no solo una), se usa `axis=1`.

```python
estudiantes.apply(
    lambda fila: f"{fila['nombre']}: {fila['calificacion']}",
    axis=1
)
```

## Cierre: glosario de la sesión

| Término | Significado |
|---|---|
| `Series` | Arreglo de NumPy con índice (etiquetas) para cada valor |
| `DataFrame` | Colección de Series que comparten el mismo índice de filas |
| `.loc[]` | Acceso por etiqueta del índice |
| `.iloc[]` | Acceso por posición numérica |
| `.head()` | Muestra las primeras filas de un DataFrame |
| `.shape` | Tupla (filas, columnas) |
| `.dtypes` | Tipo de dato de cada columna |
| `read_csv()` | Lee un archivo de texto delimitado y regresa un DataFrame |
| `to_csv()` | Escribe un DataFrame a un archivo de texto delimitado |
| `Series.map()` | Aplica una función a cada valor de una Series |
| `DataFrame.apply()` | Aplica una función a lo largo de un eje (columnas o filas) |
| `axis=0` | La función recibe una columna completa a la vez (por defecto) |
| `axis=1` | La función recibe una fila completa a la vez |
