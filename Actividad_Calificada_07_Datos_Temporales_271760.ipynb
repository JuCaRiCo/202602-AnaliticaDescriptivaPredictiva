
### 2.3 Velocidad promedio (6 pts)

La velocidad promedio de un viaje, en millas por hora, es `distance / (duracion_min / 60)`.

1. Crea la columna `velocidad_mph` **solo para viajes con duracion mayor que cero**; en los demás debe quedar como faltante (`NaN`). Sugerencia: calcula la velocidad para todos los viajes y conserva solo los valores válidos con `.where(condicion)`, que deja `NaN` donde la condicion es `False`.

## Actividad 3. Componentes y patrones (25 puntos)

### 3.1 Extraer componentes (6 pts)

Crea las columnas:

- `hora`: hora de `pickup`.
- `dia_num`: día de la semana de `pickup` (lunes = 0).
- `dia_nombre`: nombre del día de `pickup`.
- `fin_semana`: `True` si es sábado o domingo.
- `fecha`: solo la fecha (sin hora) de `pickup`.
- `cruza_medianoche`: `True` si la **fecha** de `dropoff` es distinta de la de `pickup`.

Cuenta cuántos viajes cruzan la medianoche.

```python
# Creacion columna Hora
taxis['hora'] = taxis['pickup'].dt.hour
taxis['dia_num'] = taxis['pickup'].dt.dayofweek
taxis['dia_nombre'] = taxis['pickup'].dt.day_name()
taxis['fin_semana'] = taxis['dia_num'] >= 5
taxis['fecha'] = taxis['pickup'].dt.date
taxis['cruza_medianoche'] = taxis['fecha'] != taxis['fecha'].shift()
print('Viajes que cruzan la medianoche:', taxis['cruza_medianoche'].sum())
colsNuevas = ['hora', 'dia_num', 'dia_nombre', 'fin_semana', 'fecha', 'cruza_medianoche']
print('Nuevas Columnas:\n', taxis[colsNuevas].head(5))
```

### 3.2 Patron por hora del día (7 pts)

Con `groupby('hora')`, calcula para cada hora: el **número de viajes** y la **mediana** de `duracion_min`. Indica la hora con más viajes y la hora con menos.

Usa la mediana (no la media) porque la duracion tiene valores extremos.

```python
# Agrupando y calculando
patronHora = taxis.groupby('hora').agg(
    num_viajes=('pickup', 'size'),
    mediana_duracion_min=('duracion_min', 'median')
)


# Identificar la hora con mayor y menor volumen de viajes
horaMasViajes = patronHora['num_viajes'].idxmax()
maxViajes = patronHora['num_viajes'].max()

horaMenosViajes = patronHora['num_viajes'].idxmin()
minViajes = patronHora['num_viajes'].min()

print('hora del día.:\n', patronHora)

print(f"\nHora con MÁS viajes:  {horaMasViajes}:00 hrs con {maxViajes} viajes.")
print(f"Hora con MENOS viajes: {horaMenosViajes}:00 hrs con {minViajes} viajes.")
```

### 3.3 Patron por día de la semana (6 pts)

Calcula por día de la semana el **número de viajes** y la **mediana** de `duracion_min` y de `fare`. Presenta los días en orden de lunes a domingo (sugerencia: agrupa por `['dia_num', 'dia_nombre']`).

Toma en cuenta que marzo de 2019 no tiene el mismo número de cada día de la semana: calcula también el **promedio de viajes por fecha**, dividiendo entre el número de fechas distintas de ese día de la semana (usa la columna `fecha` y `.nunique()`).

```python
# Agrupar por 'dia_num' y 'dia_nombre'
# Calculos: total de viajes, días unicos en el mes, mediana de duracion y mediana de tarifa (fare)
patronDia = taxis.groupby(['dia_num', 'dia_nombre']).agg(
    total_viajes=('pickup', 'size'),
    dias_unicos=('fecha', 'nunique'),
    mediana_duracion_min=('duracion_min', 'median'),
    mediana_fare=('fare', 'median')
)

# Calculo del promedio de viajes diarios para cada dia de la semana
patronDia['promedio_viajes_diario'] = patronDia['total_viajes'] / patronDia['dias_unicos']

print("Viajes por día de la semana (Lunes a Domingo)")
patronDia
```

## Actividad 4. Dataset limpio y comparacion (20 puntos)

### 4.1 Dataset limpio (6 pts)

Crea `taxis_limpio` aplicando el criterio que definiste en 2.4. Imprime cuántos viajes se excluyeron por cada condicion y en total, y verifica que `taxis_limpio` cumple todas las condiciones.

```python
# Criterio 1: Excluir pickups fuera de marzo de 2019
fueraMarzo = ~((taxis['pickup'].dt.year == 2019) & (taxis['pickup'].dt.month == 3))

# Criterio 2: Excluir viajes sin recorrido físico o sin tiempo (distancia = 0 O duracion = 0)
sinTrayecto = (taxis['distance'] == 0) | (taxis['duracion_min'] == 0)

# Criterio 3: Excluir velocidades inviables en entorno urbano (> 80 mph)
velocidadExtrema = taxis['velocidad_mph'] > 80

# Contar viajes excluidos por cada criterio
excluidosMarzo = fueraMarzo.sum()
excluidosTrayecto = sinTrayecto.sum()
excluidosVelocidad = velocidadExtrema.sum()

# Combinar las mascara para obtener la mascara global de exclusion
excluirTotal = fueraMarzo | sinTrayecto | velocidadExtrema
totalExcluidos = excluirTotal.sum()

# Crear el DataFrame 'taxis_limpio' aplicando la inversion de la mascara de exclusion (~mask)
taxis_limpio = taxis[~excluirTotal].copy()

# Imprimir de resultados de la limpieza
print(f"Viajes fuera de marzo 2019:                {excluidosMarzo}")
print(f"Viajes sin trayecto (distancia 0 o dur 0):  {excluidosVelocidad}")
print(f"Viajes con velocidad > 80 mph:             {excluidosVelocidad}")
print(f"TOTAL de viajes únicos excluidos:          {totalExcluidos}")
print(f"Registros originales:                     {len(taxis)}")
print(f"Registros en 'taxis_limpio':              {len(taxis_limpio)}")

# 5. Verificaciones finales sobre 'taxis_limpio'
checkFueraMarzo = (~((taxis_limpio['pickup'].dt.year == 2019) & (taxis_limpio['pickup'].dt.month == 3))).sum()
checkDistCero = (taxis_limpio['distance'] == 0).sum()
checkDuracionCero = (taxis_limpio['duracion_min'] == 0).sum()
checkVelExtrema = (taxis_limpio['velocidad_mph'] > 80).sum()

print("\nVerificacion de 'taxis_limpio' ---")
print(f"Registros fuera de marzo 2019 restante:   {checkFueraMarzo}")
print(f"Registros con distancia = 0 restante:      {checkDistCero}")
print(f"Registros con duracion = 0 restante:       {checkDuracionCero}")
print(f"Registros con velocidad > 80 mph restante: {checkVelExtrema}")
```

### 4.2 Comparacion entre periodos (6 pts)

Con `taxis_limpio`, compara los viajes que inician entre las **7:00 y las 9:59** en **días hábiles** contra los del **mismo horario en fin de semana**. Para cada grupo reporta: número de viajes, mediana de `duracion_min` y mediana de `velocidad_mph`.

```python
# Se define la mascara para la franja horaria de la mañana (7:00 a 9:59 hrs)
manana = taxis_limpio['hora'].between(7, 9)

# Se Separa entre Dias Habiles (Lunes a Viernes: fin_semana == False) y Fin de Semana (Sabado y Domingo: fin_semana == True)
habiles = manana & (~taxis_limpio['fin_semana'])
finSemana = manana & (taxis_limpio['fin_semana'])

# Extraer subconjuntos de datos
viajesMananaHabiles = taxis_limpio[habiles]
viajesMananaFinSemana = taxis_limpio[finSemana]

# Construir la tabla comparativa con las metricas solicitadas
comparativa_manana = pd.DataFrame({
    'Días Habiles (Lun-Vie)': [
        len(viajesMananaHabiles),
        viajesMananaHabiles['duracion_min'].median(),
        viajesMananaHabiles['velocidad_mph'].median()
    ],
    'Fin de Semana (Sab-Dom)': [
        len(viajesMananaFinSemana),
        viajesMananaFinSemana['duracion_min'].median(),
        viajesMananaFinSemana['velocidad_mph'].median()
    ]
}, index=['Numero de viajes', 'Mediana duracion (min)', 'Mediana velocidad (mph)'])

print("Comparativa de viajes (07:00 a 09:59 hrs) ---")
print(comparativa_manana.round(2))
```
