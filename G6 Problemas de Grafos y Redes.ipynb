```python
import networkx as nx

def crear_red_ami():
    # 1. Instanciación del Grafo
    G = nx.Graph()

    # 2. Definición de Nodos y Enlaces Ponderados (Latencia en ms)
    enlaces = [
        ("DCU", "M1", 5),  ("DCU", "M2", 8),
        ("M1",  "M3", 12), ("M1",  "M4", 15),
        ("M2",  "M4", 10), ("M2",  "M5", 22),
        ("M3",  "M4", 18), ("M4",  "M6", 8),
        ("M4",  "M5", 25), ("M5",  "M7", 7),
        ("M6",  "M8", 14), ("M7",  "M9", 11),
        ("M8",  "M9", 9)
    ]

    for u, v, w in enlaces:
        G.add_edge(u, v, weight=w)

    return G

def analizar_red(G):
    print(">>> ANÁLISIS DE LA RED SMART GRID (AMI) <<<")

    # A. Ruta Óptima (Dijkstra) desde el nodo más distante (M9) hacia el DCU
    nodo_origen = "M9"
    nodo_destino = "DCU"
    ruta_corta = nx.dijkstra_path(G, source=nodo_origen, target=nodo_destino, weight="weight")
    latencia_total = nx.dijkstra_path_length(G, source=nodo_origen, target=nodo_destino, weight="weight")

    print(f"1. RUTA ÓPTIMA DE TRANSMISIÓN ({nodo_origen} -> {nodo_destino}):")
    print(f"   Ruta: {' -> '.join(ruta_corta)}")
    print(f"   Latencia Total Acumulada: {latencia_total} ms\n")

    # B. Centralidad de Intermediación (Betweenness Centrality)
    betweenness = nx.betweenness_centrality(G, weight="weight")
    top_nodos = sorted(betweenness.items(), key=lambda x: x[1], reverse=True)

    print("2. CENTRALIDAD DE INTERMEDIACIÓN (Nodos Críticos):")
    for nodo, score in top_nodos[:5]:
        print(f"   - Nodo {nodo:4s}: Score = {score:.4f}")
    print()

    # C. Detección de Puentes (Aristas Críticas)
    puentes = list(nx.bridges(G))
    print("3. DIAGNÓSTICO DE VULNERABILIDAD (Puentes):")
    if puentes:
        print(f"   ¡Alerta! Enlaces críticos cuya falla desconecta la red: {puentes}")
    else:
        print("   La red es altamente redundante (sin puentes únicos).")

    return ruta_corta

# Ejecución del Script
red_ami = crear_red_ami()
ruta_optima = analizar_red(red_ami)
```
